```python
import os
import pandas as pd
```

```python
def get_standard_label(folder_name):
    """Standardizes the messy folder names into the 6 target categories."""
    fn = folder_name.lower()
    if 'normal' in fn: return 'Noise'
    if 'corona' in fn: return 'Corona'
    if 'turn_to_turn' in fn or 'turn to turn' in fn: return 'Turn_to_Turn'
    if 'floating' in fn: return 'Floating'
    if 'surface' in fn: return 'Surface'
    if 'void' in fn: return 'Void'
    return None
```

```python
def count_samples(root_folders):
    # Dictionary to store counts: {BigFolder: {StandardLabel: Count}}
    stats = {root: {label: 0 for label in ['Noise', 'Corona', 'Turn_to_Turn', 'Floating', 'Surface', 'Void']} 
             for root in root_folders}

    for root in root_folders:
        if not os.path.exists(root):
            print(f"Warning: Folder {root} not found.")
            continue
            
        for subfolder in os.listdir(root):
            label = get_standard_label(subfolder)
            
            if label:
                subfolder_path = os.path.join(root, subfolder)
                if os.path.isdir(subfolder_path):
                    # Count only the .csv files
                    csv_count = len([f for f in os.listdir(subfolder_path) if f.endswith('.csv')])
                    stats[root][label] += csv_count

    # Convert to a DataFrame for a nice visual table
    df = pd.DataFrame(stats).T
    # Add a 'Total' column for each big folder
    df['Total_Samples'] = df.sum(axis=1)
    
    return df
```

```python
main_folders = [
        "Data/0820", 
        "Data/0903", 
        "Data/250915_to_250916_MTR_NEW_DB_KERI"
    ]
    
report = count_samples(main_folders)

# Print the full table
print(report.to_string())

print(f"TOTAL: {report['Total_Samples'].sum()} CSV samples found.")
```
